# Travelling Salesman Problem using Dynamic Programming.
Problem Description:
Given a set of cities and distances, find the shortest possible route that visits each city once and returns to the starting city.

#Theory
TSP uses dynamic programming with bit masking. Store results of visited subsets to reduce repeated work.

## Algorithm
1. Start from source city
2. Mark visited cities using bit mask
3. Try all unvisited cities
4. Choose minimum cost path
5. Return to source city

```python
def tsp(graph):
    n = len(graph)
    dp = [[-1]*(1<<n) for _ in range(n)]

    def visit(city, mask):
        if mask == (1<<n) - 1:
            return graph[city][0]

        if dp[city][mask] != -1:
            return dp[city][mask]

        ans = float('inf')

        for next_city in range(n):
            if mask & (1<<next_city) == 0:
                new_cost = graph[city][next_city] + visit(next_city, mask | (1<<next_city))
                if new_cost < ans: n
                    ans = new_cost

        dp[city][mask] = ans
        return ans

    return visit(0, 1)
```

```python
graph = [
[0, 10, 15, 20],
[10, 0, 35, 25],
[15, 35, 0, 30],
[20, 25, 30, 0]
]

print("Minimum travelling cost:", tsp(graph))
```

```python
def tsp_with_path(graph):
    n = len(graph)
    dp_memo = [[(float('inf'), -1)] * (1 << n) for _ in range(n)]
    def visit(city, mask):
        if mask == (1 << n) - 1:
            return (graph[city][0], 0)
        if dp_memo[city][mask][0] != float('inf'):
            return dp_memo[city][mask]
        min_cost = float('inf')
        best_next_city = -1
        for next_city_idx in range(n):
            if (mask & (1 << next_city_idx)) == 0:
                cost_from_next_state, _ = visit(next_city_idx, mask | (1 << next_city_idx))
                current_total_cost = graph[city][next_city_idx] + cost_from_next_state
                if current_total_cost < min_cost:
                    min_cost = current_total_cost
                    best_next_city = next_city_idx
        dp_memo[city][mask] = (min_cost, best_next_city)
        return dp_memo[city][mask]
    final_min_cost, _ = visit(0, 1)
    path = []
    current_city = 0
    current_mask = 1
    path.append(current_city)
    while True:
        _, next_city_in_path = dp_memo[current_city][current_mask]
        if next_city_in_path == -1:
            break
        if (current_mask | (1 << next_city_in_path)) == ((1 << n) - 1):
             current_city = next_city_in_path
             path.append(current_city)
             current_mask |= (1 << current_city)
             break
        current_city = next_city_in_path
        path.append(current_city)
        current_mask |= (1 << current_city)
    path.append(0)
    return final_min_cost, path
# Take graph input from user
def get_user_graph():
    while True:
        try:
            n = int(input("Enter the number of cities: "))
            if n <= 1:
                print("Number of cities must be greater than 1.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter an integer.")
    user_graph = []
    print(f"Enter the distance matrix ({n}x{n} rows). Separate distances with spaces:")
    for i in range(n):
        while True:
            try:
                row_str = input(f"Enter row {i} (e.g., '0 10 15'): ")
                row = list(map(int, row_str.split()))
                if len(row) != n:
                    print(f"Row must contain {n} distances. Please try again.")
                else:
                    user_graph.append(row)
                    break
            except ValueError:
                print("Invalid input. Please enter integers separated by spaces.")
    return user_graph
user_defined_graph = get_user_graph()
min_cost_user, path_user = tsp_with_path(user_defined_graph)
print("\nMinimum travelling cost (with path) for user input graph:")
print(min_cost_user)
print("Path:", " -> ".join(map(str, path_user)))
```
